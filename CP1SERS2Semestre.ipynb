Uma empresa de eficiência energética está analisando o comportamento de uma residência de baixo consumo. A equipe deseja identificar períodos de consumo elevado dos eletrodomésticos e observar quais condições de temperatura e umidade estavam presentes nesses momentos.

```python
import pandas as pd
```

```python
df1 = pd.read_csv("/content/EnergydataSample.csv")
df1.head()
```

```python
df1.shape
```

```python
df1.info()
```

```python
df1.describe()
```

```python
df1.rename({"Appliances": "Consumo_Eletrodomesticos", "RH_1": "Umidade Relativa 1",
"RH_2": "Umidade Relativa 2", "RH_3": "Umidade Relativa 3"}, axis = 1, inplace = True)
df1.head()
```

```python
max = df1["Consumo_Eletrodomesticos"].max()
print(f"O maior consumo por eletrodomésticos na amostra é {max}.")
```

```python
mediaT1 = df1["T1"].mean()
df3 = df2[df2["T1"] > mediaT1]
df3.head()
```

```python
numAcima70T1 = df3["Consumo_Eletrodomesticos"].count()
percentT1 = numAcima70T1 / numAcima70
print(f"Originalmente, havia {numAcima70} registros com consumo acima de 70% do máximo.")
print(f"Após adicionar o critério de temperatura (T1 acima da média), restam {numAcima70T1} registros, o que representa {percentT1 * 100:.2f}% dos registros anteriores.")
print("\nEfeito da inclusão da temperatura como segundo critério:")
print("Ao adicionar o critério de temperatura (T1 > média de T1), o número de eventos de alto consumo de eletrodomésticos é significativamente reduzido. Isso sugere que nem todos os picos de consumo ocorrem em temperaturas ambientes mais altas (T1). O critério de temperatura atua como um filtro adicional, isolando os períodos de alto consumo que também estão associados a condições de temperatura específicas. Isso é útil para refinar a análise e focar nos eventos de consumo elevado que realmente são influenciados por temperaturas mais quentes.")
```

```python
df21 = pd.read_csv("/content/SteelIndustrySample.csv")
df21.head()
```

```python
df21.rename({"Usage_kWh": "Consumo_kWh", "Lagging_Current_Power_Factor": "LagCPF",
"Leading_Current_Power_Factor": "LeadCPF"}, axis = 1, inplace = True)
df21.head()
```

```python
df23 = df22[df22["LagCPF"] < limiarLagCPF]
df23.head()
```

```python
print("\nPor que o segundo conjunto de registros (df23) merece maior atenção da equipe de energia?\n")
print("O DataFrame `df23` contém registros onde o **Consumo_kWh** é superior a 75% do consumo máximo E o **Lagging_Current_Power_Factor (LagCPF)** está abaixo da média dos períodos de alto consumo. Essa combinação é crítica por várias razões:")
print("\n1.  **Alto Consumo (Consumo_kWh > 75% do máximo):** Isso, por si só, já indica um período de alta demanda energética. Monitorar e gerenciar esses picos é fundamental para controlar custos operacionais e garantir a estabilidade do fornecimento.")
print("\n2.  **Baixo Fator de Potência Indutivo (LagCPF < média de LagCPF):** O fator de potência é uma medida da eficiência com que a energia elétrica está sendo convertida em trabalho útil. Um baixo fator de potência (especialmente indutivo, como indicado por 'Lagging Current Power Factor'):")
print("    *   **Aumenta as Perdas:** Para a mesma quantidade de trabalho útil, um baixo fator de potência exige que a rede elétrica forneça mais corrente. Essa corrente adicional gera perdas de energia na forma de calor nos cabos e transformadores, elevando o consumo total e o desperdício.")
print("    *   **Gera Multas e Custos Adicionais:** As concessionárias de energia frequentemente aplicam multas a indústrias com baixo fator de potência, pois isso sobrecarrega a infraestrutura da rede e diminui a capacidade de fornecimento. Isso se traduz diretamente em custos mais altos na conta de energia.")
print("    *   **Reduz a Capacidade do Equipamento:** Um baixo fator de potência pode limitar a capacidade de carga real de transformadores e geradores, podendo exigir investimentos em equipamentos maiores ou correções no sistema.")
print("    *   **Impacta a Vida Útil dos Equipamentos:** O excesso de corrente pode causar aquecimento e estresse nos equipamentos, reduzindo sua vida útil e aumentando a necessidade de manutenção.")
print("\n**Conclusão:** Os registros em `df23` representam os piores cenários: não apenas há um consumo muito elevado, mas esse consumo está sendo feito de forma ineficiente, com um fator de potência desfavorável. Ao focar nesses períodos, a equipe de energia pode identificar as causas raiz (por exemplo, máquinas específicas, horários de operação, falhas de compensação de reativos) e implementar medidas corretivas para otimizar o uso de energia, reduzir custos e prolongar a vida útil dos equipamentos. Isso pode envolver a instalação de bancos de capacitores para correção do fator de potência ou a otimização de cargas indutivas.")
```

## Dataset 3 — Power Consumption of Tetouan City (UCI)

Fonte: https://archive.ics.uci.edu/dataset/849/power+consumption+of+tetouan+city

### Situação

Uma equipe responsável pelo planejamento da distribuição de energia em Tétouan deseja identificar qual das três zonas apresenta o maior pico de consumo e observar as condições ambientais registradas nos momentos de maior demanda.

untitled.svg

```python
df31 = pd.read_csv("/content/TetuanCityPowerConsumptionSample.csv")
df31.rename({"Zone 1 Power Consumption": "Consumo_Zona_1", "Zone 2  Power Consumption": "Consumo_Zona_2",
"Zone 3  Power Consumption": "Consumo_Zona_3"}, axis = 1, inplace = True)
df31.head()
```

```python
maxZ1 = df31["Consumo_Zona_1"].max()
maxZ2 = df31["Consumo_Zona_2"].max()
maxZ3 = df31["Consumo_Zona_3"].max()
print(f"O maior consumo na zona 1 é {maxZ1}, na zona 2 é {maxZ2} e na zona 3 é {maxZ3}.")
```

```python
print(f"A zona que apresenta o maior pico de consumo na amostra é a zona 1.")
```

```python
limiar70Z1 = maxZ1 * 0.7
print(f"O limiar de 70% do pico de consumo da zona 1 é {limiar70Z1}.")
df32 = df31[df31["Consumo_Zona_1"] > limiar70Z1]
df32.head()
```

```python
numAcima70Z1 = df32["Consumo_Zona_1"].count()
numTotal3 = df31["Consumo_Zona_1"].count()
percent3 = numAcima70Z1 / numTotal3
print(f"A quantidade da amostra que está acima de 70% do valor máximo da zona 1 é {numAcima70Z1}.")
print(f"O percentual da amostra que está acima de 70% do valor máximo da zona 1 é {percent3 * 100}%.")
```

```python
TMed = df31["Temperature"].mean()
print(f"A temperatura média da amostra é {TMed}.")
df33 = df32[df32["Temperature"] > TMed]
df33.head()
```

```python
numAcima70Z1Temp = df33["Consumo_Zona_1"].count()
print(f"Originalmente, havia {numAcima70Z1} registros com consumo acima de 70% do máximo na Zona 1.")
print(f"Após adicionar o critério de temperatura (Temperatura acima da média), restam {numAcima70Z1Temp} registros.")

percentual_reducao = (numAcima70Z1 - numAcima70Z1Temp) / numAcima70Z1 * 100
print(f"Isso representa uma redução de {percentual_reducao:.2f}% na quantidade de registros.\n")

print("Efeito da inclusão da condição ambiental (Temperatura > média de Temperatura):\n")
print("A inclusão do critério de temperatura (onde a temperatura é superior à média geral) resultou em uma redução significativa na quantidade de registros. Isso indica que, embora a Zona 1 apresente picos de consumo elevados em diversos momentos, uma parcela considerável desses picos não ocorre necessariamente quando a temperatura ambiente está acima da média. Ao filtrar por temperaturas mais altas, a análise se concentra especificamente nos eventos de alto consumo que estão correlacionados com condições mais quentes. Essa granularidade é valiosa para a equipe de planejamento, pois permite identificar padrões de consumo em cenários climáticos específicos, o que pode influenciar decisões sobre capacidade da rede, gerenciamento de demanda ou até mesmo estratégias de precificação em períodos de calor intenso.")
```

## Dataset 5 — Wind & Solar Energy Production (Kaggle)

Fonte: https://www.kaggle.com/datasets/ahmeduzaki/wind-and-solar-energy-production-dataset

### Situação

Um operador de um portfólio de fontes renováveis deseja comparar a ocorrência de períodos de alta produção solar e alta produção eólica. Como as duas fontes possuem escalas próprias, cada uma deverá ser comparada com o seu próprio valor máximo.

untitled2.svg

```python
df51 = pd.read_csv("/content/EnergyProductionSample.csv")
df51.loc[df51["Source"] == "Solar", "Geracao_Solar"] = df51["Production"]
df51.loc[df51["Source"] == "Wind", "Geracao_Eolica"] = df51["Production"]
df51 = df51.drop(columns = ["Source", "Production"])
df51.head()
```

```python
maxGS = df51["Geracao_Solar"].max()
maxGE = df51["Geracao_Eolica"].max()
print(f"O maior valor para geração solar é {maxGS} e para geração eólica é {maxGE}.")
```

```python
limiar70GS = maxGS * 0.7
limiar70GE = maxGE * 0.7
print(f"70% do máximo solar é {limiar70GS} e 70% do máximo eólico é {limiar70GE}.")
```

```python
df52 = df51[df51["Geracao_Solar"] > limiar70GS]
df53 = df51[df51["Geracao_Eolica"] > limiar70GE]
df52.head()
```

```python
df53.head()
```

```python
numAcima70GS = df52["Geracao_Solar"].count()
numTotal5 = df51["Geracao_Solar"].count()
percent5 = numAcima70GS / numTotal5
numAcima70GE = df53["Geracao_Eolica"].count()
numTotal6 = df51["Geracao_Eolica"].count()
percent6 = numAcima70GE / numTotal6
print(f"A quantidade da amostra que está acima de 70% do máximo solar é {numAcima70GS} e do máximo eólico é {numAcima70GE}.")
print(f"O percentual da amostra que está acima de 70% do máximo solar é {percent5 * 100}% e do máximo eólico é {percent6 * 100}%.")
print(f"A fonte que aparece com mais frequência acima de 70% de seu máximo em números absolutos e em percentual é a eólica.")
```

```python
print("Explicação sobre a inadequação de usar o mesmo valor numérico como limite para fontes de energia com escalas diferentes:")
print("\n1. Diferenças de Escala Intrínsecas:")
print("   - A geração de energia solar e eólica opera em escalas de potência muito distintas.")
print(f"   - O maior valor para geração solar (maxGS) é: {maxGS}")
print(f"   - O maior valor para geração eólica (maxGE) é: {maxGE}")
print("   - Como você pode ver, um valor absoluto (por exemplo, 15000) pode representar um pico para uma fonte, mas não para a outra.")

print("\n2. Contexto e Significado Relativo:")
print("   - O que significa 'alta produção' é relativo à capacidade máxima de cada fonte.")
print("   - Se usarmos um limite de, digamos, 15000 para ambas as fontes:")
print(f"     - Para a solar (maxGS = {maxGS}), 15000 estaria muito próximo do seu máximo, indicando altíssima produção.")
print(f"     - Para a eólica (maxGE = {maxGE}), 15000 estaria mais distante do seu máximo, sendo uma alta produção, mas não um 'pico' tão relativo.")

print("\n3. Comparação Justa e Normalização:")
print("   - Para comparar o 'alto desempenho' de forma justa entre as fontes, é essencial usar critérios relativos, como uma porcentagem da capacidade máxima de CADA UMA.")
print(f"   - Foi por isso que calculamos: 70% do máximo solar = {limiar70GS} e 70% do máximo eólico = {limiar70GE}.")
print("   - Isso permite identificar períodos de desempenho similarmente elevado em relação ao potencial de cada fonte, independentemente de suas capacidades absolutas serem diferentes.")

print("\nEm resumo, para uma análise significativa do comportamento de pico de diferentes fontes de energia, é crucial normalizar ou usar limites proporcionais à sua capacidade máxima individual, evitando interpretações errôneas.")
```
